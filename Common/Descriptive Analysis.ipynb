### Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load dataset (https://www.kaggle.com/datasets/akalyasubramanian/dataset-for-descriptive-analysis/)

```python
df = pd.read_csv("../Source data/Concrete_Data.csv")
```

###  Quick overview

```python
print(df.head())
print("...............................................")
print(df.info())
print("...............................................")
print(df.shape)  
```

```python
print(df.describe(include='all')) #Summary statistics
```

### most frequent value (Mode)

```python
print(df.mode().iloc[0]) 
```

### missing values

```python
print(df.isnull().sum())  #Missing values check
```

### Outlier Detection

```python
sns.boxplot(x=df['CMS'])  # boxplot for strength - Concrete compressive strength
plt.title("Outliers in Concrete Strength")
plt.show()
```

### Distribution plot

```python
for col in df.select_dtypes(include=np.number).columns:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], bins=30, kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
df.hist(figsize=(12,8))
plt.show()
```

### Correlation heatmap

```python
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.show()
""
```
